```python
说明：- 本文件用于演示「JUnit 框架」的测试写法与断言风格。- 采用 Python 的 unittest 模块（标准库），其 API 与 JUnit 高度对应：      @Test            -> 以 test_ 开头的方法      Assertions       -> self.assert* 系列      @BeforeEach      -> setUp()      @AfterEach       -> tearDown()      @BeforeAll       -> setUpClass()      @AfterAll        -> tearDownClass()      assertThrows     -> self.assertRaises(...)      @ParameterizedTest -> subTest()（参数化用例）"""
```

```python
import unittest
```

```python
def classify_triangle(a: float, b: float, c: float) -> str:    """判断三条边构成的三角形类型。
```

```python
    返回：        "非三角形"   — 不能构成三角形（含非正数或两边之和不大于第三边）        "等边三角形" — 三边相等        "等腰三角形" — 恰有两边相等        "一般三角形" — 三边互不相等且能构成三角形    """    # 非法输入：边长必须为正数    if a <= 0 or b <= 0 or c <= 0:        return "非三角形"
```

```python
    # 三角形成立条件：任意两边之和大于第三边    if a + b <= c or a + c <= b or b + c <= a:        return "非三角形"
```

```python
    # 类型判定    if a == b == c:        return "等边三角形"    if a == b or b == c or a == c:        return "等腰三角形"    return "一般三角形"
```

```python
class TestTriangleClassification(unittest.TestCase):    """对应 JUnit 的测试类，一个类一个被测单元。"""
```

```python
    # ---- 对应 JUnit @BeforeAll：整个测试类运行前执行一次 ----    @classmethod    def setUpClass(cls):        cls.class_run_count = 0        print("[setUpClass] 测试类初始化完成")
```

```python
    # ---- 对应 JUnit @BeforeEach：每个测试方法运行前执行 ----    def setUp(self):        TestTriangleClassification.class_run_count += 1        print(f"[setUp] 准备第 {self.class_run_count} 个用例的测试环境")
```

```python
    # ---- 对应 JUnit @AfterEach：每个测试方法运行后执行 ----    def tearDown(self):        print("[tearDown] 清理测试环境")
```

```python
    # ---- 对应 JUnit @AfterAll：整个测试类运行结束后执行一次 ----    @classmethod    def tearDownClass(cls):        print(f"[tearDownClass] 共运行 {cls.class_run_count} 个用例")
```

```python
    # ============ 等边三角形 ============    def test_equilateral_triangle(self):        """等边：三边相等。"""        self.assertEqual(classify_triangle(5, 5, 5), "等边三角形")
```

```python
    def test_equilateral_with_decimal(self):        """等边：支持浮点数。"""        self.assertEqual(classify_triangle(2.5, 2.5, 2.5), "等边三角形")
```

```python
    # ============ 等腰三角形 ============    def test_isosceles_triangle(self):        """等腰：恰有两边相等（a==b）。"""        self.assertEqual(classify_triangle(3, 3, 4), "等腰三角形")
```

```python
    def test_isosceles_triangle_other_side(self):        """等腰：恰有两边相等（b==c）。"""        self.assertEqual(classify_triangle(4, 5, 5), "等腰三角形")
```

```python
    def test_isosceles_triangle_another_side(self):        """等腰：恰有两边相等（a==c）。"""        self.assertEqual(classify_triangle(6, 4, 6), "等腰三角形")
```

```python
    # ============ 一般三角形 ============    def test_scalene_triangle(self):        """一般：三边互不相等。"""        self.assertEqual(classify_triangle(3, 4, 5), "一般三角形")
```

```python
    def test_scalene_with_decimal(self):        """一般：浮点边长。"""        self.assertEqual(classify_triangle(3.1, 4.2, 5.3), "一般三角形")
```

```python
    # ============ 非三角形 ============    def test_not_triangle_sum_too_small(self):        """非三角形：两边之和等于第三边。"""        self.assertEqual(classify_triangle(1, 2, 3), "非三角形")
```

```python
    def test_not_triangle_sum_less(self):        """非三角形：两边之和小于第三边。"""        self.assertEqual(classify_triangle(1, 1, 3), "非三角形")
```

```python
    def test_not_triangle_zero_side(self):        """非三角形：边长为 0。"""        self.assertEqual(classify_triangle(0, 4, 5), "非三角形")
```

```python
    def test_not_triangle_negative_side(self):        """非三角形：边长为负数。"""        self.assertEqual(classify_triangle(-1, 4, 5), "非三角形")
```

```python
    # ============ 其他断言风格（对应 JUnit 的 assertTrue / assertFalse / assertIn） ============    def test_result_is_valid_category(self):        """返回结果必须属于四种合法类型之一。"""        valid = {"等边三角形", "等腰三角形", "一般三角形", "非三角形"}        self.assertIn(classify_triangle(3, 4, 5), valid)
```

```python
    def test_result_not_none(self):        """返回结果不能为 None。"""        self.assertIsNotNone(classify_triangle(1, 1, 1))
```

```python
    # ============ 异常处理（对应 JUnit assertThrows） ============    def test_type_error_on_string_input(self):        """传入字符串类型应抛出 TypeError。"""        with self.assertRaises(TypeError):            classify_triangle("3", 4, 5)  # noqa
```

```python
    # ============ 参数化用例（对应 JUnit @ParameterizedTest） ============    def test_parameterized_cases(self):        """用 subTest 批量验证多组输入。"""        cases = [            ((5, 5, 5), "等边三角形"),            ((3, 3, 4), "等腰三角形"),            ((3, 4, 5), "一般三角形"),            ((1, 2, 3), "非三角形"),            ((0, 0, 0), "非三角形"),        ]        for (a, b, c), expected in cases:            with self.subTest(a=a, b=b, c=c):                self.assertEqual(classify_triangle(a, b, c), expected)
```

```python
if __name__ == "__main__":    # verbosity=2 会逐个显示用例名与结果，类似 JUnit 的报告输出    unittest.main(verbosity=2)
```
